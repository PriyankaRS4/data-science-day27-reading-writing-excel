# Reading and Writing Excel Data
Task 27 using the Superstore Dataset.

```python
import pandas as pd

df = pd.read_excel("superstore_input.xlsx", sheet_name="Superstore")
print("Shape:", df.shape)
df.head()
```

```python
print("Columns:", df.columns.tolist())
print("\nData types:\n", df.dtypes)
print("\nMissing values:\n", df.isna().sum())
```

```python
cleaned = df.copy()
cleaned.columns = [str(c).strip() for c in cleaned.columns]
cleaned = cleaned.dropna(how='all').drop_duplicates()
for col in cleaned.select_dtypes(include='object').columns:
    cleaned[col] = cleaned[col].apply(lambda x: x.strip() if isinstance(x, str) else x)
for col in ["Sales", "Profit", "Quantity", "Discount"]:
    if col in cleaned.columns:
        cleaned[col] = pd.to_numeric(cleaned[col], errors='coerce')
print("Cleaned shape:", cleaned.shape)
cleaned.head()
```

```python
cleaned.to_excel("superstore_processed.xlsx", sheet_name="Processed_Data", index=False)
print("Exported: superstore_processed.xlsx")
```
